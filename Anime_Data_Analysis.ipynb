## Rating Distribution

### Objective
The objective of this visualization is to analyze the distribution of anime ratings in the dataset. Understanding the rating distribution helps identify whether most anime receive high, average, or low ratings from viewers.

```python
sns.histplot(df["Rating"], kde=True)

plt.title("Anime Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Anime")

plt.show()
```

### Observation
- The majority of anime ratings are concentrated between 7 and 8.
- Very few anime have extremely low or extremely high ratings.
- The distribution indicates that most anime are positively rated by viewers.

### Insight
The dataset suggests that anime generally receive favorable audience ratings, indicating a high overall quality of content.

## Distribution of Anime Types

### Objective
The purpose of this visualization is to examine the number of anime available in different categories such as TV Series, Movies, OVAs, Specials, and ONAs.

```python
sns.countplot(x="Type", data=df)
plt.show()
```

### Observation
- TV anime has the highest number of entries.
- Movies and OVAs appear less frequently.
- Specials and ONAs represent a smaller portion of the dataset.

### Insight
TV series dominate the anime industry, making them the most common production format.

## Top 10 Highest Rated Anime

### Objective
This visualization identifies the ten highest-rated anime in the dataset based on user ratings.

```python
top10 = df.sort_values("Rating", ascending=False).head(10)

sns.barplot(x="Rating", y="Name", data=top10)
plt.show()
```

### Observation
- These anime have the highest ratings among all titles.
- The difference in ratings among the top-ranked anime is relatively small.
- These titles represent the most highly appreciated anime by viewers.

### Insight
Highly rated anime can be considered benchmark titles and may reflect strong storytelling, animation quality, and audience satisfaction.

## Distribution of Episodes

### Objective
The purpose of this graph is to understand how the number of episodes varies among anime in the dataset.

```python
sns.histplot(df["Episodes"])
plt.show()
```

### Observation
- Most anime contain a relatively small number of episodes.
- A few anime have a significantly larger episode count.
- The distribution is right-skewed due to long-running series.

### Insight
Short and medium-length anime are much more common than long-running productions.

## Correlation Heatmap

### Objective
The objective of this heatmap is to examine the relationships between numerical variables in the dataset.

```python
sns.heatmap(df.corr(numeric_only=True),
            annot=True)

plt.show()
```

### Observation
- Darker colors indicate stronger relationships between variables.
- Values closer to 1 represent strong positive correlations.
- Values near 0 indicate weak or no relationship.

### Insight
The heatmap helps identify which numerical features are related and supports deeper data analysis.

## Top Anime Studios

### Objective
This visualization identifies the studios that have produced the highest number of anime in the dataset.

```python
top_studios = df["Studio"].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_studios.values, y=top_studios.index)
plt.title("Top 10 Anime Studios")
plt.xlabel("Number of Anime")
plt.ylabel("Studio")
plt.show()
```

Top 10 Studios

### Observation
- A small number of studios contribute a large share of anime production.
- Several studios have relatively fewer productions.

### Insight
Anime production is concentrated among a few major studios that consistently produce large numbers of titles.

## Most Common Anime Genres

### Objective
The objective of this visualization is to identify the genres that appear most frequently in the dataset.

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="Release_season")
plt.title("Anime Released by Season")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(df["Episodes"], bins=30)
plt.title("Episodes Distribution")
plt.show()
```

## Anime Release Trend Over the Years

### Objective
This visualization analyzes how anime production has changed across different years.

```python
year = df["Release_year"].value_counts().sort_index()

plt.figure(figsize=(12,5))
plt.plot(year.index, year.values, marker="o")
plt.title("Anime Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Anime")
plt.grid(True)
plt.show()
```

```python
tags = df["Tags"].dropna().str.split(",")

tag_list = []

for item in tags:
    for tag in item:
        tag_list.append(tag.strip())

top_tags = pd.Series(tag_list).value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_tags.values, y=top_tags.index)
plt.title("Top 10 Anime Tags")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

# Key Insights

2. Most anime ratings are between 7 and 8.

3. Anime production increased significantly after 2000.

4. A few studios produce a large share of anime.

5. Most anime have fewer than 30 episodes.
